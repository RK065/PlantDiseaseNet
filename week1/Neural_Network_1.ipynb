```python
import numpy as np
```

```python
#Data
X = np.array([
    [1.0, 2.0, 1.0, 2.0],
    [2.0, 1.0, 2.0, 1.0],
    [1.5, 2.5, 1.0, 2.5],

    [5.0, 4.0, 5.0, 4.0],
    [4.5, 5.0, 4.0, 5.0],
    [5.5, 4.5, 5.0, 4.5],

    [8.0, 1.0, 8.0, 1.0],
    [7.5, 1.5, 7.0, 1.5],
    [8.5, 1.0, 8.5, 1.0],

    [2.0, 8.0, 2.0, 8.0],
    [1.5, 7.5, 2.0, 7.0],
    [2.5, 8.5, 2.5, 8.0]
])

y = np.array([
    0, 0, 0,
    1, 1, 1,
    2, 2, 2,
    3, 3, 3
])
```

```python
# Parameters
np.random.seed(42)

W1 = np.random.randn(4, 8) * 0.1
b1 = np.zeros((1, 8))

W2 = np.random.randn(8, 6) * 0.1
b2 = np.zeros((1, 6))

W3 = np.random.randn(6, 4) * 0.1
b3 = np.zeros((1, 4))
```

```python
# Activation functions
def relu(z):
    return np.maximum(0, z)


def softmax(z):
    z = z - np.max(z, axis=1, keepdims=True)
    exp_z = np.exp(z)
    return exp_z / np.sum(exp_z, axis=1, keepdims=True)
```

```python
# Forward pass
def forward(X):
    Z1 = X @ W1 + b1
    A1 = relu(Z1)

    Z2 = A1 @ W2 + b2
    A2 = relu(Z2)

    Z3 = A2 @ W3 + b3
    probabilities = softmax(Z3)

    return Z1, A1, Z2, A2, Z3, probabilities
```

```python
# Loss
def cross_entropy(probabilities, y):
    m = len(y)

    correct_probabilities = probabilities[
        np.arange(m), y
    ]

    return -np.mean(
        np.log(correct_probabilities + 1e-8)
    )
```

```python
# Backpropagation
def backward(X, y, Z1, A1, Z2, A2, probabilities):

    m = len(X)

    dZ3 = probabilities.copy()
    dZ3[np.arange(m), y] -= 1
    dZ3 /= m

    dW3 = A2.T @ dZ3
    db3 = np.sum(dZ3, axis=0, keepdims=True)

    dA2 = dZ3 @ W3.T
    dZ2 = dA2 * (Z2 > 0)

    dW2 = A1.T @ dZ2
    db2 = np.sum(dZ2, axis=0, keepdims=True)

    dA1 = dZ2 @ W2.T
    dZ1 = dA1 * (Z1 > 0)

    dW1 = X.T @ dZ1
    db1 = np.sum(dZ1, axis=0, keepdims=True)

    return dW1, db1, dW2, db2, dW3, db3
```

```python
# Training
learning_rate = 0.05
epochs = 5000

for epoch in range(epochs):

    Z1, A1, Z2, A2, Z3, probabilities = forward(X)

    loss = cross_entropy(probabilities, y)

    dW1, db1, dW2, db2, dW3, db3 = backward(
        X, y, Z1, A1, Z2, A2, probabilities
    )

    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1

    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2

    W3 -= learning_rate * dW3
    b3 -= learning_rate * db3

    if epoch % 500 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.4f}")
```

```python
# Predictions
_, _, _, _, _, probabilities = forward(X)

predictions = np.argmax(probabilities, axis=1)

accuracy = np.mean(predictions == y)

print("\nPredictions:")
print(predictions)

print("\nActual:")
print(y)

print(f"\nAccuracy: {accuracy * 100:.2f}%")
```
