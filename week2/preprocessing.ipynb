```python
import os
from collections import Counter

import matplotlib.pyplot as plt
from PIL import Image
```

```python
dataset_path = "../data/PlantVillage"

print("Dataset Path:", dataset_path)
print("Classes Found:")
print(os.listdir(dataset_path))
```

```python
class_counts = {}

for class_name in os.listdir(dataset_path):
    class_path = os.path.join(dataset_path, class_name)

    if os.path.isdir(class_path):
        image_count = len(os.listdir(class_path))
        class_counts[class_name] = image_count

for class_name, count in sorted(class_counts.items()):
    print(f"{class_name}: {count}")
```

```python
total_images = sum(class_counts.values())

print("Total Images:", total_images)
print("Number of Classes:", len(class_counts))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14,6))

plt.bar(class_counts.keys(), class_counts.values())

plt.xticks(rotation=90)

plt.xlabel("Disease Classes")
plt.ylabel("Number of Images")
plt.title("PlantVillage Dataset Distribution")

plt.tight_layout()
plt.show()
```

Dataset Summary

- Dataset: PlantVillage
- Total Images: 20,639
- Total Classes: 15

The dataset contains images of healthy and diseased leaves from tomato, potato, and pepper plants.

Each class corresponds to a specific disease condition or healthy plant state.

The dataset will be used to train a deep learning model for plant disease classification.

```python
from PIL import Image
import matplotlib.pyplot as plt
import os

fig, axes = plt.subplots(3, 3, figsize=(12, 12))

classes = sorted(os.listdir(dataset_path))

for ax, class_name in zip(axes.flatten(), classes[:9]):

    class_folder = os.path.join(dataset_path, class_name)

    image_file = os.listdir(class_folder)[0]

    image_path = os.path.join(class_folder, image_file)

    img = Image.open(image_path)

    ax.imshow(img)
    ax.set_title(class_name[:25], fontsize=8)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
image_sizes = []

for class_name in sorted(os.listdir(dataset_path)):

    class_folder = os.path.join(dataset_path, class_name)

    image_file = os.listdir(class_folder)[0]

    image_path = os.path.join(class_folder, image_file)

    img = Image.open(image_path)

    image_sizes.append((class_name, img.size))

for cls, size in image_sizes:
    print(cls, "->", size)
```

```python
total_size = 20639

train_size = int(0.8 * total_size)
val_size = int(0.1 * total_size)
test_size = total_size - train_size - val_size

print("Train:", train_size)
print("Validation:", val_size)
print("Test:", test_size)
```

## Dataset Splitting Strategy

The dataset is divided into three subsets:

### Training Set (80%)

Used for learning model parameters.

### Validation Set (10%)

Used for:

- Hyperparameter tuning
- Model selection
- Monitoring overfitting

### Test Set (10%)

Used only after training is complete.

This provides an unbiased estimate of real-world model performance.

Split Ratio:

- Train: 80%
- Validation: 10%
- Test: 10%
